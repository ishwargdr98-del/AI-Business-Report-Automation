```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")

print("Dataset Shape:", df.shape)
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
missing_count = df.isna().sum()

missing_percentage = (missing_count / len(df)) * 100


missing_report = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

display(missing_report)
```

```python
print("Total Rows:", len(df))

print("CustomerID Missing:",
      df["CustomerID"].isnull().sum())

print("CustomerID Available:",
      df["CustomerID"].notnull().sum())
```

```python
print("Missing Description:",
      df["Description"].isnull().sum())

print("\nRows with Missing Description:")
display(df[df["Description"].isnull()].head())
```

```python
print("Zero UnitPrice rows:",
      (df["UnitPrice"] == 0).sum())

print("Negative Quantity rows:",
      (df["Quantity"] < 0).sum())

print("Negative UnitPrice rows:",
      (df["UnitPrice"] < 0).sum())
```

```python
# Negative quantity वाले रिकॉर्ड देखना
negative_quantity = df[df["Quantity"] < 0]

print("Negative Quantity Records:")
display(negative_quantity.head(10))
```

```python
# Step 1: Create Data Quality Audit Report

audit_summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Missing CustomerID",
        "Missing Description",
        "Negative Quantity",
        "Zero UnitPrice",
        "Negative UnitPrice"
    ],
    "Count": [
        len(df),
        df["CustomerID"].isna().sum(),
        df["Description"].isna().sum(),
        (df["Quantity"] < 0).sum(),
        (df["UnitPrice"] == 0).sum(),
        (df["UnitPrice"] < 0).sum()
    ]
})

display(audit_summary)
```

```python
# Save audit report as CSV
audit_summary.to_csv("data_quality_audit.csv", index=False)

# Read the saved file back
saved_report = pd.read_csv("data_quality_audit.csv")

display(saved_report)
```
